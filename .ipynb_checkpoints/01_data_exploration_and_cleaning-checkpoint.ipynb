# 7CS537 CW1, Notebook 01: Data Exploration and Cleaning (CICIDS2017)

**Purpose:** load the CICIDS2017 MachineLearningCSV files, examine data quality, clean the data and save a cleaned dataset for model training (Notebook 02).

**Case study framing:** CICIDS2017 stands in for TeleComNet Ltd's network flow data. Each row is one network flow produced by CICFlowMeter, labelled as BENIGN or as a specific attack.

**Outputs saved to `outputs/`:** summary tables (CSV) and figures (PNG) for use as testing evidence in the report and presentation.

Run the cells in order. Every decision that changes the data is printed with before and after counts so it can be reported.

## 1. Setup and configuration

```python
import os
import re
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Folder containing the 8 CSV files from MachineLearningCSV.zip (MachineLearningCVE folder)
DATA_DIR = "data/MachineLearningCVE"
OUT_DIR = "outputs"
os.makedirs(OUT_DIR, exist_ok=True)

# Methodological switch: Destination Port can act as a shortcut feature
# (e.g. "port 21 traffic is FTP-Patator") that does not generalise to a real network.
# Default is to drop it; record the choice in the methodology section of the report.
DROP_DESTINATION_PORT = True

RANDOM_STATE = 42
pd.set_option("display.max_columns", 100)
print("pandas", pd.__version__)
```

## 2. Load all eight CSV files

```python
files = sorted(glob.glob(os.path.join(DATA_DIR, "*.csv")))
print(f"Found {len(files)} files")
assert len(files) > 0, "No CSV files found. Check DATA_DIR."

frames = []
file_summary = []
for f in files:
    # latin-1 avoids decode errors on the dash character in the Web Attack labels
    df_part = pd.read_csv(f, encoding="latin-1", low_memory=False)
    df_part.columns = df_part.columns.str.strip()          # column names have leading spaces
    df_part["source_file"] = os.path.basename(f)
    frames.append(df_part)
    file_summary.append({"file": os.path.basename(f), "rows": len(df_part), "columns": df_part.shape[1] - 1})

df = pd.concat(frames, ignore_index=True)
del frames

file_summary = pd.DataFrame(file_summary)
file_summary.to_csv(f"{OUT_DIR}/01_files_loaded.csv", index=False)
display(file_summary)
print("Combined shape:", df.shape)
```

## 3. Clean the label column
The Web Attack labels contain a non-standard dash character, so they are normalised to readable names.

```python
df["Label"] = df["Label"].astype(str).str.strip()

def normalise_label(lbl):
    if lbl.startswith("Web Attack"):
        if "Brute" in lbl: return "Web Attack - Brute Force"
        if "XSS" in lbl:   return "Web Attack - XSS"
        if "Sql" in lbl:   return "Web Attack - Sql Injection"
    return lbl

df["Label"] = df["Label"].map(normalise_label)
print(sorted(df["Label"].unique()))
```

## 4. Raw class distribution (before cleaning)

```python
raw_counts = df["Label"].value_counts().rename("raw_count").to_frame()
raw_counts["raw_pct"] = (raw_counts["raw_count"] / len(df) * 100).round(4)
display(raw_counts)

# Which attacks appear in which file (day of capture)
label_by_file = pd.crosstab(df["Label"], df["source_file"])
label_by_file.to_csv(f"{OUT_DIR}/01_label_by_file.csv")
display(label_by_file)
```

## 5. Data quality checks

```python
feature_cols = [c for c in df.columns if c not in ("Label", "source_file")]
num = df[feature_cols].apply(pd.to_numeric, errors="coerce")

quality = pd.DataFrame({
    "nan_count": num.isna().sum(),
    "inf_count": np.isinf(num).sum(),
    "n_unique": num.nunique(dropna=True),
})
quality = quality[(quality.nan_count > 0) | (quality.inf_count > 0) | (quality.n_unique <= 1)]
quality.to_csv(f"{OUT_DIR}/01_quality_issues.csv")
print("Columns with missing, infinite or constant values:")
display(quality)

# Columns that are exact copies of another column
dup_cols = []
seen = {}
for c in feature_cols:
    key = pd.util.hash_pandas_object(num[c], index=False).sum()
    if key in seen and num[c].equals(num[seen[key]]):
        dup_cols.append((c, seen[key]))
    else:
        seen[key] = c
print("Duplicated columns (column, copy of):", dup_cols)

dup_rows = df.duplicated(subset=feature_cols + ["Label"]).sum()
print(f"Duplicate rows: {dup_rows:,} ({dup_rows/len(df)*100:.2f}%)")

negative_cols = (num < 0).sum()
negative_cols = negative_cols[negative_cols > 0]
print("Columns containing negative values (worth discussing, e.g. window sizes of -1):")
display(negative_cols)
```

## 6. Cleaning steps
Each step logs the row or column counts so the report can state exactly what was removed and why.

```python
log = []
def record(step, before, after, unit="rows"):
    log.append({"step": step, "before": before, "after": after, "removed": before - after, "unit": unit})
    print(f"{step}: {before:,} -> {after:,} {unit} (removed {before - after:,})")

df[feature_cols] = num  # use the numeric version of the features
del num

# 6a. Infinite values become NaN, then rows with NaN are dropped
before = len(df)
df[feature_cols] = df[feature_cols].replace([np.inf, -np.inf], np.nan)
df = df.dropna(subset=feature_cols)
record("Drop rows with NaN or infinite values", before, len(df))

# 6b. Remove exact duplicate rows (prevents the same flow appearing in both train and test)
before = len(df)
df = df.drop_duplicates(subset=feature_cols + ["Label"])
record("Drop duplicate rows", before, len(df))

# 6c. Drop duplicated and constant columns
before_cols = len(feature_cols)
drop_cols = [c for c, _ in dup_cols]
drop_cols += [c for c in feature_cols if df[c].nunique() <= 1 and c not in drop_cols]
if DROP_DESTINATION_PORT and "Destination Port" in feature_cols:
    drop_cols.append("Destination Port")
df = df.drop(columns=drop_cols)
feature_cols = [c for c in feature_cols if c not in drop_cols]
print("Dropped columns:", drop_cols)
record("Drop duplicated, constant and excluded columns", before_cols, len(feature_cols), unit="columns")

# 6d. Reduce memory: float32 is enough precision for these features
df[feature_cols] = df[feature_cols].astype(np.float32)

pd.DataFrame(log).to_csv(f"{OUT_DIR}/01_cleaning_log.csv", index=False)
print("Final shape:", df.shape)
```

## 7. Target definitions
Two targets are created so Notebook 02 can choose between them:
- **`is_attack`** (binary): the main detection task, BENIGN versus any attack.
- **`attack_family`**: groups the 15 labels into families, used to break down detection performance by attack type (e.g. does the model miss rare web attacks?).

```python
family_map = {
    "BENIGN": "Benign",
    "DoS Hulk": "DoS", "DoS GoldenEye": "DoS", "DoS slowloris": "DoS", "DoS Slowhttptest": "DoS",
    "DDoS": "DDoS",
    "PortScan": "PortScan",
    "FTP-Patator": "Brute Force", "SSH-Patator": "Brute Force",
    "Web Attack - Brute Force": "Web Attack", "Web Attack - XSS": "Web Attack",
    "Web Attack - Sql Injection": "Web Attack",
    "Bot": "Botnet",
    "Infiltration": "Infiltration",
    "Heartbleed": "Heartbleed",
}
unmapped = set(df["Label"]) - set(family_map)
assert not unmapped, f"Unmapped labels: {unmapped}"

df["attack_family"] = df["Label"].map(family_map)
df["is_attack"] = (df["Label"] != "BENIGN").astype(np.int8)

clean_counts = df["Label"].value_counts().rename("clean_count").to_frame()
dist = raw_counts.join(clean_counts, how="left").fillna(0)
dist["clean_pct"] = (dist["clean_count"] / dist["clean_count"].sum() * 100).round(4)
dist["family"] = dist.index.map(family_map)
dist.to_csv(f"{OUT_DIR}/01_class_distribution.csv")
display(dist)

print(df["is_attack"].value_counts(normalize=True).rename({0: "Benign", 1: "Attack"}))
```

## 8. Visualising the class imbalance

```python
fig, ax = plt.subplots(figsize=(10, 6))
d = dist.sort_values("clean_count")
ax.barh(d.index, d["clean_count"])
ax.set_xscale("log")
ax.set_xlabel("Number of flows (log scale)")
ax.set_title("CICIDS2017 class distribution after cleaning")
for i, v in enumerate(d["clean_count"]):
    ax.text(v, i, f" {int(v):,}", va="center", fontsize=8)
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/01_class_distribution.png", dpi=200)
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
fam = df["attack_family"].value_counts()
ax.bar(fam.index, fam.values)
ax.set_yscale("log")
ax.set_ylabel("Number of flows (log scale)")
ax.set_title("Flows per attack family")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.savefig(f"{OUT_DIR}/01_attack_families.png", dpi=200)
plt.show()
```

## 9. Feature overview
A quick look at feature ranges. Very different scales are one reason scaling will matter for some models (e.g. Logistic Regression) and not others (tree-based models).

```python
desc = df[feature_cols].describe().T[["mean", "std", "min", "max"]]
desc.to_csv(f"{OUT_DIR}/01_feature_summary.csv")
display(desc.sort_values("max", ascending=False).head(15))
```

## 10. Save the cleaned dataset for Notebook 02

```python
keep = feature_cols + ["Label", "attack_family", "is_attack", "source_file"]
try:
    df[keep].to_parquet("data/cicids2017_clean.parquet", index=False)
    print("Saved data/cicids2017_clean.parquet")
except ImportError:
    df[keep].to_pickle("data/cicids2017_clean.pkl")
    print("pyarrow not installed, saved data/cicids2017_clean.pkl instead")

with open(f"{OUT_DIR}/01_feature_list.txt", "w") as fh:
    fh.write("\n".join(feature_cols))
print(f"{len(feature_cols)} features, {len(df):,} rows")
```
